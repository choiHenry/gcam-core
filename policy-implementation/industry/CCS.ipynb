```python
import pandas as pd

dfDef = pd.read_csv("../../analysis/extdata/gdpdef.csv")
dfDef.head()
dfDef['Year'] = dfDef['date'].str.split('-').str[0].astype(int)
arrDef = dfDef.set_index('Year')['gdpdef']
arrDef[2024] = (139.71495 / 136.41462) * arrDef[2023]
def gcam_deflator(value, from_year=2023, to_year=1975):
    return value * arrDef[to_year] / arrDef[from_year]

dfExc = pd.read_csv("../../analysis/extdata/DEXKOUS.csv")
dfExc['date'] = pd.to_datetime(dfExc['observation_date'])
dfExcAnnual = dfExc.groupby(dfExc['date'].dt.year)['DEXKOUS'].mean()

def krw_to_usd(year):
    return 1 / dfExcAnnual[year]
```

# CCS Subsidy

The Act on the Capture, Transportation, Storage, and Utilization of Carbon Dioxide (CCUS Act) of Korea provides the legal basis for financial, administrative, and technical support for CCS/CCUS projects, including funding for demonstration projects, R&D subsidies, concessional loans, and investment through the Climate Response Fund. However, the Act does not stipulate a specific subsidy amount per tonne of CO₂ captured and stored. In contrast, the United States currently offers a tax credit of USD 84/tCO₂ for dedicated geological storage under the Inflation Reduction Act’s enhanced 45Q provision. Given the absence of an explicit Korean subsidy level, this study assumes a CCS support level of KRW 8,000/tCO2 for the *Current Policies* scenario, and that of KRW 16,000/tCO2 for the *Enhanced Ambition* scenario using the U.S. benchmark as an upper bound for policy feasibility analysis.
