# Chemical

In the *Enhanced Ambition* scenario, it is assumed that biomass-based feedstock technologies are commercialized in line with the findings of this study ([SMDI2021Lee](../resources/SMDI-2021-Lee), [Flaim1981Biomass](https://doi.org/10.2172/6335593), [usda2011_biomass_supply](https://www.usda.gov/sites/default/files/documents/biomass-supply-chain-report.pdf), [ahorsu2018_biomass_review](https://doi.org/10.3390/en11123366)). Following the ratification of the UN Plastics Treaty ([guardian2025_plastics](https://www.theguardian.com/environment/2025/aug/05/un-plastic-pollution-talks-must-result-in-ambitious-treaty-leading-expert-says), [apnews2025_geneva](https://apnews.com/article/615d096e211daa1593c7d0f16745f1ff)), the plastic recycling rate is projected to increase to 25% by 2040. This assumption reflects the effect of chemical recycling—where waste plastics are converted into petrochemical feedstocks—by proportionally reducing the feedstock input coefficient of refined liquids in the chemical industry.

These technology assumptions are excluded from the *Current Policies* scenario for two main reasons. First, the UN Plastics Treaty has not yet been concluded, and thus no binding domestic implementation measures exist in Korea ([guardian2025_plastics](https://www.theguardian.com/environment/2025/aug/05/un-plastic-pollution-talks-must-result-in-ambitious-treaty-leading-expert-says), [apnews2025_geneva](https://apnews.com/article/615d096e211daa1593c7d0f16745f1ff)). Consequently, there is no policy certainty to support a significant increase in the national plastic recycling rate. Second, the commercialization of biomass-based feedstock technologies in the petrochemical industry faces considerable technical and economic challenges ([Flaim1981Biomass](https://doi.org/10.2172/6335593), [usda2011_biomass_supply](https://www.usda.gov/sites/default/files/documents/biomass-supply-chain-report.pdf), [ahorsu2018_biomass_review](https://doi.org/10.3390/en11123366)). The production of biomass-derived feedstocks at scale requires reliable supply chains, proven conversion technologies, and competitive costs relative to fossil-based feedstocks, none of which are currently secured in Korea. As such, both measures are considered beyond the scope of current legislated and implemented policies.
