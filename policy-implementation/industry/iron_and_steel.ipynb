# Iron and Steel

First of all, we calibrated the model's 2020 iron and steel production in 2020 to Korean production data (KSIS-2023-MOTIE) focusing on the fact that in 2020, the import of iron and steel is declined drastically in Korea: `/input/policy/korea-2035/industry/ironsteel_import_constraint_value.xml`

Next, in line with the Carbon Neutrality Technology Development Program for the Four Major Industries, we assume that hydrogen-based DRI is commercialized starting in 2035, at which point the share-weight of the H2REX technology is set to 1: `/input/policy/korea-2035/industry/iron_and_steel_korea.xml`

We assume that the use of scrap in steelmaking remains at its current level: `/input/policy/korea-2035/industry/ironsteel_scrap_constraint_value_cp`. In addition, we assume that regulations on extending the lifetime of blast furnaces are gradually tightened, but a complete prohibition on lifetime extensions is not implemented until 2040: `/input/policy/korea-2035/industry/iron_steel_cp.xml`

In the *Enhanced Ambition* scenario, we assume that the ban of the lifetime extensions is implemented from 2030: `/input/policy/korea-2035/industry/iron_steel_cp.xml` and the use of scrap increases at 30%/5-year: `/input/policy/korea-2035/industry/ironsteel_scrap_constraint_value_cp`
