# Cement

In line with the Carbon Neutrality Technology Development Program for the Four Major Industries, we assume that limestone calcined clay cement (LC3) technology will be commercialized starting in 2030. The detailed implementation process is outlined below.

1. **LC³ blend composition** [@shao2025_lc3; @rmi2023_lc3; @hu2024_lc3; @maraghechi2018_lc3]

   * Clinker: 50 % of final cement mass  
   * Calcined clay: 30 % (omitted here)  
   * Additional limestone: 15 %

2. **Limestone needed for clinker production**

   * Portland clinker contains ~65 % CaO by mass.  
   * To produce 0.65 t CaO you need:  
     \[
     0.65 \times \frac{100.09}{56.08} \approx 1.16 \ \text{t CaCO}_3
     \]
   * At 50 % clinker share:  
     \[
     1.16 \times 0.50 = \mathbf{0.58\ t\ limestone\ per\ t\ cement}
     \]

3. **Separate limestone component**

   * LC³ blend also requires an extra **0.15 t** limestone per t cement.

4. **Total limestone coefficient**

   \[
     0.58\ (\text{for clinker}) + 0.15\ (\text{blend addition}) = 0.73
   \]
