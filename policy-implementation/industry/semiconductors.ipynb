# Semiconductors

Samsung Electronics has announced plans to replace CF₄ and SF₆ with alternative gases by 2027, as reported by [ET News](https://www.etnews.com/20230517000206). In this study, we assume that CF₄ and SF₆ are fully replaced starting in 2030 (see `/input/policy/korea-2035/industry/semiconductors_cp.xml`). In the *Enhanced Ambition* scenario, we further assume that C₂F₆ and HFC-23 are also phased out in semiconductor manufacturing processes (see `/input/policy/korea-2035/industry/semiconductors_ep.xml`).
