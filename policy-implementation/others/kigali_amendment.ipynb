# Kigali Amendment

South Korea has ratified the *Kigali Amendment* to the Montreal Protocol, committing to the phasedown of hydrofluorocarbons (HFCs).  
However, policy instruments to achieve the targeted HFC reductions remain insufficient.  

In the *Current Policy* scenario, we assume that no additional measures are introduced beyond existing domestic regulations to fulfill the Kigali Amendment.  
In the *Enhanced Ambition* scenario, we assume the introduction of an HFC tax—converted using GWP values from AR5—at a level consistent with the projected EU ETS carbon prices, calculated following the methodology in `../industry/carbon_prices.ipynb`.
