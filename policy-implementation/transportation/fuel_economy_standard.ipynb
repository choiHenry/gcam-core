# Fuel Economy Standard

According to the Administrative Notice on Average Greenhouse Gas Emission and Fuel Economy Standards for Automobiles (2021–2030), Ministry of Environment, Korea’s passenger cars are required to improve fuel economy by 30% by 2030 relative to 2020 levels. In the *Enhanced Ambition* scenario, we assume a 45% improvement by 2035. For freight trucks, we assume a 30% improvement from 2025 to 2035. To maintain consistency with the model, the fuel economy improvement rates were applied to GCAM’s default values to create the input file.
