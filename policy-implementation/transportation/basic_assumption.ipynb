# Performance per Vehicle by Mode

# Car (pass-km/veh)

* daily milage (vkt / veh / day) : 33 [TMACS](https://tmacs.kotsa.or.kr/web/TG/TG200/TG2200/Tg1700_02.jsp?mid=S3080#)
* lifespan (year) : 15.3 [KADRA](http://www.tbnews.co.kr/news/view.php?idx=1784)
* load factor (pass/veh): 1.26 [KTDB](https://www.ktdb.go.kr/common/downloadFile.do?fileNm=/PassengerTripSurvey/2009_press01.pdf&dnFileNm=%EC%84%9C%EC%9A%B8%EC%8B%9C%20%EC%B6%9C%EA%B7%BC%EC%8B%9C%EA%B0%84%EB%8C%80%20%EB%82%98%ED%99%80%EB%A1%9C%20%EC%B0%A8%EB%9F%89%EB%B9%84%EC%9C%A8%2086.3%ED%8D%BC%EC%84%BC%ED%8A%B8.pdf)

$\text{performance (pass-km/veh)}=\text{daily milage (km/day/veh)} * 365 * \text{lifespan (year)} * \text{load factor (pass/veh)}$

```python
33 * 365
```

```python
# UCD pass-km / veh
pass_km_per_veh = 33 * 365 * 15.3 * 1.26
pass_km_per_veh
```

# Bus (pass-km/veh)

* Total performance per year (pass-km / year) in 2022: 70,241,290,694 [KTDB](https://www.ktdb.go.kr/www/selectTrnsportTreeView.do?key=32)
* Total number of bus enrolled(veh) in 2020: 85,165 [MOLIT](https://stat.molit.go.kr/portal/cate/statView.do?hRsId=313&hFormId=&hDivEng=&month_yn=)
* Lifespan (year) : 15.3 [KADRA](http://www.tbnews.co.kr/news/view.php?idx=1784)

$\text{performance (pass-km/veh)}=\frac{\text{total yearly performance (pass-km/year)} * 365 * \text{lifespan (year)}}{\text{total number of buses enrolled (veh)}}$

```python
70241290694 / 85165
```

```python
pass_km_per_veh = 70241290694 / 85165 * 15.5
pass_km_per_veh
```

# Medium Truck ton-km

```python
ton_km_per_veh = 47 * 365 * 16.8 * 4.2
ton_km_per_veh
```
