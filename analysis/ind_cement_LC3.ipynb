1. **LC³ blend composition**

   * Clinker: 50 % of final cement mass
   * Calcined clay: 30 % (omitted here)
   * Additional limestone: 15 %

2. **Limestone needed for clinker production**

   * Portland clinker contains \~65 % CaO by mass.
   * To produce 0.65 t CaO you need 0.65 × (100.09 / 56.08) ≈ 1.16 t CaCO₃.
   * At 50 % clinker share, raw limestone for clinker = 1.16 × 0.50 = **0.58 t limestone per t cement**.

3. **Separate limestone component**

   * LC³ blend also requires an extra 0.15 t limestone per t cement.

4. **Total limestone coefficient**

   $$
     0.58\;(\text{for clinker})\;+\;0.15\;(\text{blend addition})
     \;=\;0.73
   $$

   → Set `<minicam-energy-input name="limestone"><coefficient>0.73</coefficient></…>` so that “limestone” alone accounts both for clinker feedstock and the additional SCM fraction.
