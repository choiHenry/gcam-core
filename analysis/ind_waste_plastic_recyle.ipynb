# Enhanced Policy

요청하신 세 가지 조합

1. **S₁:** (회수율 r=0.18, 열분해=0, 가스화=0)
2. **S₂:** (r=0.30, 열분해=0.50, 가스화=0.45, 기계 0.05)
3. **S₃:** (r=0.35, 열분해=0.50, 가스화=0.45, 기계 0.05; 단 가스화 수율 55%)

을 동일한 계산 프레임으로 비교했습니다. (다른 파라미터는 ‘중간’ 가정 유지)

### 공통 파라미터(명시 안 된 것은 모두 고정)

| 항목             | 값                                                                              |
| -------------- | ------------------------------------------------------------------------------ |
| 기계 재활용 수율/품질   | $ \alpha^{M}=0.65,\ \eta^{M}=0.90$                                             |
| 열분해 수율/품질      | S₂: $ \alpha^{Py}=0.60,\ \eta^{Py}=0.90$; S₃ 동일                                |
| 가스화 수율/품질      | S₂: $ \alpha^{G}=0.50,\ \eta^{G}=0.85$; S₃: $ \alpha^{G}=0.55,\ \eta^{G}=0.85$ |
| 감쇄(다운사이클·리바운드) | $ B=0.06$                                                                      |
| 기호             | $f_{Py}, f_G, f_M$ = 회수물 중 경로 비중                                               |

### 공식

$$
\text{유효 대체 합( per 회수율 )} = \sum_{i} f_i \alpha^i \eta^i
$$

$$
D = r \big( \text{유효 대체 합} - B \big), \qquad k^{vir}=1-D
$$

$k^{vir}$: 절대 버진 refined liquids 투입 계수.

### 4. 해석 포인트

1. **회수율 상승 자체 + 화학재활용 경로 도입**(S₁→S₂) 효과가 첫 번째 큰 절감(약 3.8%)을 만든 뒤,
2. **회수율 추가 상승(30→35%) + 가스화 수율 향상(0.50→0.55)**(S₂→S₃) 이 두 번째(추가 3.2%) 절감을 제공.
3. 세 시나리오 간 총 격차는 **약 7%p 버진 계수 감소**(0.906 → 0.843) 수준.

> 요약: “(0.18,0,0) → (0.30,0.5,0.45) → (0.35,0.6,0.55)” 전환은 버진 석유화학 feedstock 투입을 단계적으로 **약 −3.5% → 추가 −3% → 누적 약 −7%**(수요 10 Mt 기준 0.35 Mt + 0.28 Mt ≈ 0.63 Mt) 줄이는 시나리오로 해석할 수 있습니다.

### 5. 주의/선택 사항

* **S₁에서 f\_M=1**로 둔 것은 “초기에는 화학재활용이 전무, 회수된 것은 전부 기계재활용 가능” 가정입니다. 실제로는 미회수·선별손실·소각이 존재하므로 S₁의 계수(0.9055)는 다소 ‘낙관적’일 수 있습니다.

  * 더 보수적으로 S₁에서 f\_M<1 (나머지 손실)로 두면 S₁의 유효 대체 합이 내려가며 S₂, S₃ 대비 절감 폭(%)은 *더 커* 보이게 됩니다. 필요하면 대안도 계산해 드릴 수 있습니다.
* 감쇄계수 $B$ (0.06) 변화는 세 시나리오 모두 선형으로 영향을 주므로, 다른 B 값에 대한 빠른 보정은 $D_r$ 에서 $r \times \Delta B$ 만큼을 더하거나 빼면 됩니다.
