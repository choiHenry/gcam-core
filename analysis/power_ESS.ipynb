In the 11th Basic plan, Korea plans to supply 10.6 TWh in 2030, 32.5 TWh in 2038 with Energy Storage Units, etc. We assume these are supplied by PV Storage and Wind Storage. The target values(10.6, 32.5) are assigned proportionately to the actual generation in 2023.

```python
# dictCapTWh = {2020: 6.3 * solar_share, 2023: 7.6 * solar_share, 2030: 10.6 * solar_share, 2038: 32.5 * solar_share}
# dictCapTWh = {2023: 7.6 * solar_share, 2030: 10.6 * solar_share, 2038: 32.5 * solar_share}
dictCapTWh = {2020: 3.271, 2023: 3.784, 2030: 11.8, 2035: 28.5}
dictCapTWh[2025] = dictCapTWh[2023] + (dictCapTWh[2030] - dictCapTWh[2023]) * (2/7)
dictCapTWh
```
