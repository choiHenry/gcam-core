# Hydro

* Data Source:
    * the 11th Basic Plan for Supply and Demand of Power(in Korean, "제11차전력수급기본계획")
    * 2023 Electricity Statistics of Korea(in Korean, "2023년 한국전력통계(2024년도판, 제93호)")

* Conventional Hydro: Korea have generated and is planning to generate 3.7 TWh / year
* Pumped Hydro: included in `Others (기타)` of `Projection of Generation Mix (발전량 전망)`. As of 2023, pumped hydro generated 3.8 TWh and `Others` generated 8.3 thus Pumped Hydro takes

```python
3.8 / 8.3
```

* 45% of `Others`. We assume this share is maintained in the future.

* Projected Generation of `Others` (TWh):
    * 2025: 9.3
    * 2030: 11.8
    * 2035: 28.5

```python
9.3 * (3.8/8.3), 11.8 * (3.8 / 8.3), 28.5 * 3.8 / 8.3
```

* Predicted Generation of Pumped Hydro:

    * 2030: 5.4
