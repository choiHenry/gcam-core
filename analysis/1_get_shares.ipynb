# Emissions

```python
import pandas as pd
```

```python
dfMap = pd.read_excel("./extdata/rice50_gcam71_mappings.xlsx")
dfMap.head()
```

```python
# check total CO2 emissions by country for 2018 in the EDGAR 2023 data
dfEdgar = pd.read_excel("./extdata/EDGARv8.0_FT2022_GHG_booklet_2023.xlsx", sheet_name='GHG_totals_by_country')
dfEdgar
```

```python
# Open World Bank Data
dfWorldBank = pd.read_csv("./extdata/API_EN.GHG.CO2.MT.CE.AR5_DS2_en_csv_v2_15981.csv", skiprows=[0, 1, 2, 3])
dfWorldBank.head()
```

```python
# Find missing values of EDGAR 2023 data from World Bank data

# Prepare EDGAR data
dfLeft = dfEdgar[['EDGAR Country Code', 'Country', 2018]].iloc[:210]
# Prepare World Bank data
dfRight = dfWorldBank[['Country Name', 'Country Code', '2018']]
# Merge two data sets
dfM = dfLeft.merge(dfRight, left_on='EDGAR Country Code', right_on='Country Code', how='outer')

# Find missing values of EDGAR 2023 data from World Bank data
dfM['emiss_co2_2018_mtc'] = dfM[2018].fillna(dfM['2018'] / 3.66667)

# Abbreviate Country Code Variable by CC
dfM['CC'] = dfM['EDGAR Country Code'].fillna(dfM['Country Code'])
dfM[['CC', 'emiss_co2_2018_mtc']]
```

```python
# Compare mapping data and emissions data 
lstCheck = []
for country in dfMap['Country ISO Code'].unique():
    if country not in dfM['CC'].unique():
        lstCheck.append(country)
print(lstCheck)
```

```python
dfM['Country ISO Code'] = dfM['CC']
```

```python
# Merge Map data and emissions data
dfMapEmiss = dfMap.merge(dfM[['Country ISO Code', 'emiss_co2_2018_mtc']], on='Country ISO Code', how='left')
dfMapEmiss
```

```python
dfMapEmiss[dfMapEmiss['emiss_co2_2018_mtc'].isna()]
```

# GDP/Population

```python
# Open WEO data
dfWeo = pd.read_csv("./extdata/WEOOct2024all.csv", encoding='windows-1252', thousands=',')
dfWeo.head()
```

```python
dfGdPop = dfWeo[dfWeo['WEO Subject Code'].isin(['PPPGDP', 'LP'])].copy()
dfGdPop
```

```python
lstCheck = []
for country in dfMap['Country ISO Code'].unique():
    if country not in dfGdPop['ISO'].unique():
        lstCheck.append(country)
print(lstCheck)
```

```python
dfGdPop['Country ISO Code'] = dfGdPop['ISO']
```

```python
dfGdp = dfGdPop[dfGdPop['WEO Subject Code'] == 'PPPGDP'].copy()
dfGdp['pppgdp_2018'] = dfGdp['2018']
dfGdp.head()
```

```python
dfGdp.columns
```

```python
dfMapEmissGdp = dfMapEmiss.merge(dfGdp[['Country ISO Code', 'pppgdp_2018']], on='Country ISO Code', how='left')
dfMapEmissGdp
```

```python
dfMapEmissGdp[dfMapEmissGdp['pppgdp_2018'].isna()]
```

```python
dfPop = dfGdPop[dfGdPop['WEO Subject Code'] == 'LP'].copy()
dfPop['pop_2018'] = dfPop['2018']
dfPop.head()
```

```python
dfMapEmissGdPop = dfMapEmissGdp.merge(dfPop[['Country ISO Code', 'pop_2018']], on='Country ISO Code', how='left')
dfMapEmissGdPop
```

```python
dfMapEmissGdPop[
    (dfMapEmissGdPop['emiss_co2_2018_mtc'].isna()) | (dfMapEmissGdPop['pppgdp_2018'].isna()) | (dfMapEmissGdPop['pop_2018'].isna())
]
```

```python
dfMapEmissGdPop['emiss_co2_2018_mtc'] = dfMapEmissGdPop['emiss_co2_2018_mtc'].fillna(0)
dfMapEmissGdPop['pppgdp_2018'] = dfMapEmissGdPop['pppgdp_2018'].fillna(0)
dfMapEmissGdPop['pop_2018'] = dfMapEmissGdPop['pop_2018'].fillna(0)
dfMapEmissGdPop
```

```python
dfMapEmissGdPop['pppgdp_2018'] = dfMapEmissGdPop['pppgdp_2018'].replace(",", '', regex=True).astype(float)
dfMapEmissGdPop['pop_2018'] = dfMapEmissGdPop['pop_2018'].replace(",", '', regex=True).astype(float)
dfMapEmissGdPop
```

```python
# Group the DataFrame by the 'group_column' and calculate the sum for each group
group_sums = dfMapEmissGdPop.groupby('RICE 50+')['pppgdp_2018'].transform('sum')

# Calculate the share of each numeric value within its group
dfMapEmissGdPop['s_y_RICE'] = dfMapEmissGdPop['pppgdp_2018'] / group_sums

# Group the DataFrame by the 'group_column' and calculate the sum for each group
group_sums = dfMapEmissGdPop.groupby('RICE 50+')['pop_2018'].transform('sum')

# Calculate the share of each numeric value within its group
dfMapEmissGdPop['s_p_RICE'] = dfMapEmissGdPop['pop_2018'] / group_sums

# Group the DataFrame by the 'group_column' and calculate the sum for each group
group_sums = dfMapEmissGdPop.groupby('RICE 50+')['emiss_co2_2018_mtc'].transform('sum')

# Calculate the share of each numeric value within its group
dfMapEmissGdPop['s_e_RICE'] = dfMapEmissGdPop['emiss_co2_2018_mtc'] / group_sums

# Group the DataFrame by the 'group_column' and calculate the sum for each group
group_sums = dfMapEmissGdPop.groupby('GCAM v7.1')['pppgdp_2018'].transform('sum')

# Calculate the share of each numeric value within its group
dfMapEmissGdPop['s_y_GCAM'] = dfMapEmissGdPop['pppgdp_2018'] / group_sums

# Group the DataFrame by the 'group_column' and calculate the sum for each group
group_sums = dfMapEmissGdPop.groupby('GCAM v7.1')['pop_2018'].transform('sum')

# Calculate the share of each numeric value within its group
dfMapEmissGdPop['s_p_GCAM'] = dfMapEmissGdPop['pop_2018'] / group_sums

# Group the DataFrame by the 'group_column' and calculate the sum for each group
group_sums = dfMapEmissGdPop.groupby('GCAM v7.1')['emiss_co2_2018_mtc'].transform('sum')

# Calculate the share of each numeric value within its group
dfMapEmissGdPop['s_e_GCAM'] = dfMapEmissGdPop['emiss_co2_2018_mtc'] / group_sums

dfMapEmissGdPop
```
