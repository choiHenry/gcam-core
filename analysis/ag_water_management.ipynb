# Water Management

Ministry of Agriculture, Food and Rural Affairs plans to subsidize ID(Intermittently Drainage, 간단관개) and CF+WS(Continuously Flooding+Water-Saving, 논물얕게대기) to attain adoption shares of 61.1% and 10%  each.

According to [Kim et al. (2013)](../../extdata/Effects_of_Water_Management_Methods_on_CH4_and_N2O.pdf), ID and CF+WS decreases CH4 and increases N2O, resulting in decreaseing GWP compared to CF.

```python
# Create the DataFrame
data = {
    'Practice': ['WS', 'CF+WS', 'ID', 'CF (Baseline)'],
    'CH4_Emissions_kg_per_ha': [60.8, 81.6, 194.9, 263.0],
    'N2O_Emissions_kg_per_ha': [0.019, 0.014, 0.009, 0.003],
    'CH4_Reduction_vs_CF_%': [-76.9, -69.0, -25.9, 0],
    'N2O_Increase_vs_CF_%': [533, 367, 200, 0],
}

df = pd.DataFrame(data)
df['GWP'] = [1.32, 1.85, 4.51, 6.02]
df['GWP']/= 6.02
df['GWP'] *= 100
df
```
